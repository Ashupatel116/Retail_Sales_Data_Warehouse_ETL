```python
print("\n----- DATE RANGE -----")
print("Minimum Order Date:", df_clean["order_date"].min())
print("Maximum Order Date:", df_clean["order_date"].max())
```

```python
date_range = pd.date_range(
    start=df_clean["order_date"].min(),
    end=df_clean["order_date"].max(),
    freq="D"
)
```

```python
dim_date = pd.DataFrame({
    "full_date": date_range
})
```

```python
dim_date["date_key"] = (
    dim_date["full_date"].dt.year * 10000
    + dim_date["full_date"].dt.month * 100
    + dim_date["full_date"].dt.day
)

dim_date["day"] = dim_date["full_date"].dt.day

dim_date["month"] = dim_date["full_date"].dt.month

dim_date["month_name"] = dim_date["full_date"].dt.month_name()

dim_date["quarter"] = "Q" + dim_date["full_date"].dt.quarter.astype(str)

dim_date["year"] = dim_date["full_date"].dt.year
```

```python
print("\n----- DATE DIMENSION -----")
print(dim_date.head())
print("\nRows:", len(dim_date))
```

```python
dim_customer = dim_customer.reset_index(drop=True)

dim_customer["customer_key"] = (
    dim_customer.index + 1
)
```

```python
dim_location = dim_location.reset_index(drop=True)

dim_location["location_key"] = (
    dim_location.index + 1
)

dim_location = dim_location[
    [
        "location_key",
        "country",
        "city",
        "state",
        "postal_code",
        "region"
    ]
]
```

```python
fact_sales = df_clean.merge(
    dim_customer[["customer_key", "customer_id"]],
    on="customer_id",
    how="left"
)
```

```python
fact_sales = fact_sales.merge(
    dim_location[
        [
            "location_key",
            "country",
            "city",
            "state",
            "postal_code",
            "region"
        ]
    ],
    on=[
        "country",
        "city",
        "state",
        "postal_code",
        "region"
    ],
    how="left"
)
```

```python
fact_sales = fact_sales.merge(
    dim_date[["date_key", "full_date"]],
    left_on="order_date",
    right_on="full_date",
    how="left"
)
```

```python
product_check = (
    df_clean
    .groupby("product_id")
    .agg(
        product_names=("product_name", "nunique"),
        categories=("category", "nunique"),
        sub_categories=("sub_category", "nunique")
    )
)

print(
    product_check[
        (product_check["product_names"] > 1) |
        (product_check["categories"] > 1) |
        (product_check["sub_categories"] > 1)
    ]
)
```

```python


product_check = (
    df_clean
    .groupby("product_id")
    .agg(
        product_names=("product_name", "nunique"),
        categories=("category", "nunique"),
        sub_categories=("sub_category", "nunique")
    )
)


problem_products = product_check[
    (product_check["product_names"] > 1) |
    (product_check["categories"] > 1) |
    (product_check["sub_categories"] > 1)
]

# Get the problematic product IDs
problem_ids = problem_products.index.tolist()

# Show the actual conflicting records
problem_product_details = (
    df_clean[
        df_clean["product_id"].isin(problem_ids)
    ][
        ["product_id", "product_name", "category", "sub_category"]
    ]
    .drop_duplicates()
    .sort_values("product_id")
)

print("Problematic Product IDs:", len(problem_products))

print("\n----- PROBLEMATIC PRODUCTS -----")
print(problem_product_details.to_string(index=False))
```
